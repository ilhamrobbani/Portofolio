# Data Cleansing

```python
from scipy import stats
import numpy as np
import pandas as pd

# Pastikan nama dataframe asal sesuai, di sini diasumsikan 'df'
df_clean = df.copy()

# 1. Membuat kolom date dengan format datetime yang aman
df_clean['date'] = pd.to_datetime(
    df_clean['order_date']).dt.normalize()  # normalize menghilangkan komponen jam

# 2. Menghapus semua baris tanpa customer_id
df_clean = df_clean[~df_clean['customer_id'].isna()]

# 3. Mengkonversi customer_id menjadi string (dan hapus .0 jika tipenya float)
df_clean['customer_id'] = df_clean['customer_id'].astype(
    float).astype(int).astype(str)

# 4. Menghapus semua baris tanpa product_name
df_clean = df_clean[~df_clean['product_name'].isna()]

# 5. Membuat semua product_name berhuruf kecil
df_clean['product_name'] = df_clean['product_name'].str.lower()

# 6. Menghapus semua baris dengan product_code atau product_name mengandung 'test' (Logika diperbaiki dengan &)
df_clean = df_clean[(~df_clean['product_code'].str.lower().str.contains('test', na=False)) &
                    (~df_clean['product_name'].str.contains('test', na=False))]

# 7. Menghapus baris dengan status cancelled, yaitu yang order_id-nya diawali 'C'
df_clean = df_clean[df_clean['order_id'].astype(str).str[:1] != 'C']

# 8. Mengubah nilai quantity yang negatif menjadi positif
df_clean['quantity'] = df_clean['quantity'].abs()

# 9. Menghapus baris dengan price bernilai negatif
df_clean = df_clean[df_clean['price'] > 0]

# 10. Membuat nilai amount, yaitu perkalian antara quantity dan price
df_clean['amount'] = df_clean['quantity'] * df_clean['price']

# 11. Mengganti product_name dengan yang paling sering muncul berdasarkan product_code
most_freq_product_name = df_clean.groupby(['product_code', 'product_name'], as_index=False).agg(
    order_cnt=('order_id', 'nunique')).sort_values(['product_code', 'order_cnt'], ascending=[True, False])
most_freq_product_name['rank'] = most_freq_product_name.groupby(
    'product_code')['order_cnt'].rank(method='first', ascending=False)
most_freq_product_name = most_freq_product_name[most_freq_product_name['rank'] == 1].drop(columns=[
                                                                                         'order_cnt', 'rank'])

df_clean = df_clean.merge(most_freq_product_name.rename(
    columns={'product_name': 'most_freq_product_name'}), how='left', on='product_code')
df_clean['product_name'] = df_clean['most_freq_product_name']
df_clean = df_clean.drop(columns='most_freq_product_name')

# 12. Menghapus outlier menggunakan Z-Score pada kolom numerik
df_clean = df_clean[(np.abs(stats.zscore(
    df_clean[['quantity', 'amount']])) < 3).all(axis=1)]
df_clean = df_clean.reset_index(drop=True)

# Tampilkan hasil akhir
df_clean
```

# Menyiapkan Data Basket

# Buat DataFrame Basket

```python
basket = pd.pivot_table(df_clean, index='order_id', columns='product_name', values='product_code', aggfunc='nunique', fill_value=0)
basket
```

```python
basket.info()
```

# Encode DataFrame basket dengan nilai True untuk semua nilai di atas 0 dan False untuk semua nilai 0

```python
def encode(x):
    if x==0:
        return False
    if x>0:
        return True

basket_encode = basket.applymap(encode)
basket_encode
```

```python
basket_encode.info()
```

# Ambil transaksi dengan banyaknya produk unik lebih dari 1 saja

```python
basket_filter = basket_encode[(basket_encode>0).sum(axis=1)>1]
basket_filter
```

```python
basket_filter.info()
```

# Mengaplikasikan apriori algorithm

# Buat list frequent itemset (kumpulan produk yang sering dibeli

```python
# pip install mlxtend
from mlxtend.frequent_patterns import apriori

frequent_itemset = apriori(basket_filter, min_support=.01, use_colnames=True).sort_values('support', ascending=False).reset_index(drop=True)
frequent_itemset['product_cnt'] = frequent_itemset['itemsets'].apply(lambda x: len(x))
frequent_itemset
```

# Hitung nilai support, confidence, dan lift dari setiap pasangan produk yang mungkin

```python
from mlxtend.frequent_patterns import association_rules

product_association = association_rules(frequent_itemset, metric='confidence', min_threshold=.7).sort_values(['support','confidence'], ascending=[False,False]).reset_index(drop=True)
product_association
```
